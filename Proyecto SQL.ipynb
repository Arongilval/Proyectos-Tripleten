# Encuentra el número de reseñas de usuarios y la calificación promedio para cada libro. 

```python
query = """
SELECT
    b.book_id,
    b.title,
    COALESCE(rv.num_reviews, 0) AS num_reviews,
    rt.avg_rating
FROM books b
LEFT JOIN (
    SELECT book_id, COUNT(*) AS num_reviews
    FROM reviews
    GROUP BY book_id
) rv USING (book_id)
LEFT JOIN (
    SELECT book_id, ROUND(AVG(rating), 2) AS avg_rating
    FROM ratings
    GROUP BY book_id
) rt USING (book_id)
ORDER BY num_reviews DESC NULLS LAST, avg_rating DESC NULLS LAST;
"""
```

```python
df_books_reviews = pd.read_sql(query, con=engine)
```

# Identifica la editorial que ha publicado el mayor número de libros con más de 50 páginas (esto te ayudará a excluir folletos y publicaciones similares de tu análisis).

```python
q_top1 = """
SELECT
    p.publisher,
    COUNT(*) AS books_over_50_pages
FROM publishers p
JOIN books b ON b.publisher_id = p.publisher_id
WHERE b.num_pages > 50
GROUP BY p.publisher
ORDER BY books_over_50_pages DESC
LIMIT 1;
"""
```

```python
edi_top1 = pd.read_sql(q_top1, con=engine)
```

# Identifica al autor que tiene la más alta calificación promedio del libro: mira solo los libros con al menos 50 calificaciones.

```python
query = """
WITH book_stats AS (
    SELECT
        b.book_id,
        b.author_id,
        COUNT(r.rating_id) AS n_ratings,
        AVG(r.rating)      AS avg_rating
    FROM books b
    JOIN ratings r ON r.book_id = b.book_id
    GROUP BY b.book_id, b.author_id
),
qualified AS (
    SELECT * 
    FROM book_stats
    WHERE n_ratings >= 50
)
SELECT
    a.author,
    ROUND(SUM(q.avg_rating * q.n_ratings) / SUM(q.n_ratings), 3) AS author_weighted_avg_rating,
    SUM(q.n_ratings) AS total_ratings_considered,
    COUNT(*)         AS qualifying_books
FROM qualified q
JOIN authors a ON a.author_id = q.author_id
GROUP BY a.author
ORDER BY author_weighted_avg_rating DESC
LIMIT 1;
"""
```

```python
best_author = pd.read_sql(query, con=engine)
```

```python
best_author
```

La autora mas valorada y mejor calificada es JK Rowling autora de la saga Harry potter

Rowling es la autora mejor valorada con un promedio ponderado de 4.287. </div>


# Encuentra el número promedio de reseñas de texto entre los usuarios que calificaron más de 50 libros.

```python
query = """
WITH raters AS (
    SELECT username, COUNT(*) AS ratings_count
    FROM ratings
    GROUP BY username
    HAVING COUNT(*) > 50
),
user_reviews AS (
    SELECT username, COUNT(*) AS reviews_count
    FROM reviews
    GROUP BY username
)
SELECT ROUND(AVG(COALESCE(ur.reviews_count, 0)), 2) AS avg_text_reviews_among_power_raters
FROM raters r
LEFT JOIN user_reviews ur USING (username);
"""
```

```python
df = pd.read_sql(query, con=engine)
```

```python
df
```

Hay un promedio de 24.33 reseñas que hacen usuarios que dieron a mas de 50 libros su calificacion.

## Agregar datos por usuario

lo que hice en el primer extrar el numero de llamadas por mes con group by 

```python
# Calcula el número de llamadas hechas por cada usuario al mes. Guarda el resultado.


calls_per_user_per_month = calls.groupby(['user_id', 'month']).size().reset_index(name='calls_count')
print(calls_per_user_per_month.head(10))
```

```python
# Calcula la cantidad de minutos usados por cada usuario al mes. Guarda el resultado.
minutes_per_user_per_month = calls.groupby(['user_id', 'month'])['duration'].sum().reset_index()
minutes_per_user_per_month.rename(columns={'duration': 'minutes_used'}, inplace=True)
print(minutes_per_user_per_month.head(10))
```

```python
# Calcula el número de mensajes enviados por cada usuario al mes. Guarda el resultado.

messages_per_user_per_month = messages.groupby(['user_id', 'month']).size().reset_index(name='messages_count')
print(messages_per_user_per_month.head(10))
```

```python
# Calcula el volumen del tráfico de Internet usado por cada usuario al mes. Guarda el resultado.

internet_per_user_per_month = internet.groupby(['user_id', 'month'])['gb_used'].sum().reset_index()
print(internet_per_user_per_month.head(10))
```

<div class="alert alert-block alert-success">
<b>Comentario del revisor</b> <a class="tocSkip"></a>

Perfecto, buen trabajo estructurando cada una de las tablas agrupadas. 
</div>

se utilizo la herramienta group by para extrar la cantidades de llamadas, gb usados y mensajes utilizados por mes

```python
# Fusiona los datos de llamadas, minutos, mensajes e Internet con base en user_id y month
client_info = pd.merge(calls_per_user_per_month, minutes_per_user_per_month, on=['user_id', 'month'], how='outer')
client_info = pd.merge(client_info, messages_per_user_per_month, on=['user_id', 'month'], how='outer')
client_info = pd.merge(client_info, internet_per_user_per_month, on=['user_id', 'month'], how='outer')
print(client_info.head(5))
```

```python
# Añade la información de la tarifa
client_info = client_info.merge(users[['user_id', 'plan']], on='user_id', how='left')
client_info = client_info.merge(plans, how='left', left_on='plan', right_on='plan_name')
client_info = client_info.drop(columns=['plan_name'])
```

```python
client_info = pd.merge(client_info, users[['user_id', 'city']], on='user_id', how='left')
client_info = client_info.fillna(0)
client_info
```

Perfecto, buen trabajo realizando los merges necesarios paraestructurarr la tabla con toda la información consolidada.

aqui lo que hice fue ir haciendo las cuentas agregando columnas con los resultados que necesitaba cosas interesantes a resaltar es el .clip(lower=0) que es para poner el limite inferior sea 0 y el np ceil que lo utilice para redondear los gb al limite superior mas proximo
lo primero fue sacar el minutes_over que fue para sacar la cantidad de minutos si se sobre paso o no a su limite
messages_over similar al anterior para saber si se sobre paso de su cantidad de mensaje
gb_over para poder saber si se sobrepaso en la cantidad de gb utilizados en su paquete
este proceso es para saber si se le cobrara extras dependiendo del paquete que tengan

posterior mente se hicieron cuentas con el extra_minutes_fee, extra_messages_fee y extra_gb_fee  es sacar el costo por los limites superados en cada uno de los extras dependiendo de lo que custa en usd cada extra

lo ultimo fue sumar todos los costos extras para obtener el total_revenue por cliente
cree una ultima variable que contiene el user_id, el mes, su ganancia generada y su paquete para poder tener una perspectiva visual de cuanto gasta y que paquete tiene 

```python
# Calcula el ingreso mensual para cada usuario
client_info['minutes_over'] = (client_info['minutes_used'] - client_info['minutes_included']).clip(lower=0)
client_info['messages_over'] = (client_info['messages_count'] - client_info['messages_included']).clip(lower=0)
client_info['mb_over'] = (client_info['gb_used'] - client_info['mb_per_month_included']).clip(lower=0)
```

```python
client_info['gb_over'] = (client_info['mb_over'] / 1024).apply(np.ceil)
```

```python
client_info['extra_minutes_fee'] = client_info['minutes_over'] * client_info['usd_per_minute']
client_info['extra_messages_fee'] = client_info['messages_over'] * client_info['usd_per_message']
client_info['extra_gb_fee'] = client_info['gb_over'] * client_info['usd_per_gb']
```

```python
client_info['total_revenue'] = (client_info['usd_monthly_pay'] + client_info['extra_minutes_fee'] + client_info['extra_messages_fee'] + client_info['extra_gb_fee'])
client_info = client_info.fillna(0)
```

```python
client_total_revenue = client_info[['user_id', 'month', 'total_revenue', 'plan']]
print(client_total_revenue)
```

## Prueba las hipótesis estadísticas

[Prueba la hipótesis de que son diferentes los ingresos promedio procedentes de los usuarios de los planes de llamada Ultimate y Surf.]

por lo que pudimos comprobar en la hipotesis los valores entre planes son diferentes como comente al paquete surf genera mas ganancia, se soprepasa con mas facilidad sus limites de paquete pero aun asi las actividades realizadas en ambos paquete son similares en utilizacion pero sin duda el paquete surf es mas beneficioso economicamente

```python
# Prueba las hipótesis

from scipy.stats import ttest_ind

ultimate_revenue = client_info[client_info['plan'] == 'ultimate']['total_revenue']
surf_revenue = client_info[client_info['plan'] == 'surf']['total_revenue']

t_stat, p_value = ttest_ind(ultimate_revenue, surf_revenue, equal_var=False)

print(f'Estadístico t: {t_stat:.4f}')
print(f'Valor p: {p_value:.4f}')

alpha = 0.05
if p_value < alpha:
    print('Rechazamos la hipótesis nula: hay diferencia significativa en los ingresos promedio.')
else:
    print('No se puede rechazar la hipótesis nula: no hay diferencia significativa en los ingresos promedio.')
```

[Prueba la hipótesis de que el ingreso promedio de los usuarios del área NY-NJ es diferente al de los usuarios de otras regiones.]

[Elabora las hipótesis nula y alternativa, escoge la prueba estadística, determina el valor alfa.]

```python
# Prueba las hipótesis

ny_nj_users = client_info[client_info['city'].str.contains('New York-Newark-Jersey City, NY-NJ-PA MSA', na=False)]
ny_nj_total_revenue = ny_nj_users['total_revenue'].mean()
print(f"Promedio de ganancias totales de usuarios de NY-NJ: ${ny_nj_total_revenue:,.2f}")
other_users = client_info[~client_info['city'].str.contains('New York-Newark-Jersey City, NY-NJ-PA MSA', na=False)]
other_users_total_revenue = other_users['total_revenue'].mean()
print(f"Promedio de ganancias totales de usuarios fuera de NY-NJ: ${other_users_total_revenue:,.2f}")
```

```python
from scipy import stats
alpha = 0.05
results = stats.ttest_ind(ny_nj_users['total_revenue'], other_users['total_revenue'], equal_var=False)
if results.pvalue < alpha:
    print("Rechazamos la hipótesis nula: hay una diferencia significativa en los ingresos promedio.")
else:
    print("No se puede rechazar la hipótesis nula: no hay evidencia de diferencia significativa en los ingresos.")
```
